## 1. What does the distribution of passenger ages look like?

```python
df["age"].plot(kind = "hist", bins = 15)

plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.title("Distribution of Passenger age")
plt.show()
```

## 2. What does the distribution of passenger fares look like?

```python
df["fare"].plot(kind = "hist", bins = 15)

plt.xlabel("Fare")
plt.ylabel("Number of Passengers")
plt.title("Distribution of Passenger Fares")
plt.show()
```

## 3. How does fare variation differ across passenger classes?

```python
df.groupby("pclass")["fare"].describe()
```

```python
df.boxplot(column = "fare", by = "pclass")

plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.title("Fare Variation by Passenger Class")
plt.suptitle("")
plt.show()
```

## 4. How does survival vary by passenger age?

```python
df.groupby("survived")["age"].describe()
```

```python
df.boxplot(column = "age", by = "survived")

plt.xlabel("Survival Status")
plt.ylabel("Age")
plt.title("Passenger Age Survival Status")
plt.suptitle("")
plt.show()
```

## 5. Is there a relationship between fare and age?

```python
plt.scatter(df["age"], df["fare"])

plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Relationship Between Passenger Age and Fare")
plt.show()
```

## 6. How strongly are age, fare, and survival related?

```python
correlation = df[["age", "fare", "survived"]].corr()

correlation
```

```python
plt.imshow(correlation, cmap = "seismic")

plt.xticks(range(len(correlation.columns)), correlation.columns)
plt.yticks(range(len(correlation.columns)), correlation.columns)

plt.colorbar()
plt.title("Correlation of Age, Fare, and Survival")

plt.show()
```

## 7. How does the number of family members relate to passenger fare?

```python
df["FamilySize"] = df["sibsp"] + df["parch"] + 1
```

```python
avg_family_fare = df.groupby("FamilySize")["fare"].mean()

avg_family_fare.plot(kind = "bar")

plt.xlabel("Family Size")
plt.ylabel("Average Fare")
plt.title("Average Fare by Family Size")
plt.xticks(rotation = 0)
plt.show()
```
